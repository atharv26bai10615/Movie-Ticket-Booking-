```python
# Movie Ticket Booking System
MOVIES={
    1: ("Avengers Endgame", 200),
    2: ("Spider Man No Way Home", 180),
    3: ("Interstellar", 220),
    4: ("Inception", 200),
}
SHOW_TIMES = {
    1: "10:00 AM",
    2: "2:00 PM",
    3: "6:00 PM",
    4: "9:00 PM",
}
bookings=[]
def readnumber(prompt):
    try:
        return int(input(prompt))
    except ValueError:
        print("Please enter a number.")
        return None
def showmovies():
    print("\nMovies:")
    for number, (title, price) in MOVIES.items():
        print(f"{number}. {title} - Rs. {price}")
def showtimings():
    print("\nShow Timings:")
    for number, showtime in SHOW_TIMES.items():
        print(f"{number}. {showtime}")
def bookticket():
    showmovies()
    moviechoice=readnumber("\nEnter movie number: ")
    if moviechoice not in MOVIES:
        print("Movie not available.")
        return
    moviename, price=MOVIES[moviechoice]
    showtimings()
    timechoice=readnumber("\nEnter timing number: ")
    if timechoice not in SHOW_TIMES:
        print("Timing not available.")
        return
    showtime=SHOW_TIMES[timechoice]
    ticketcount=readnumber("Enter number of tickets: ")
    if ticketcount is None or ticketcount <= 0:
        print("Please enter a valid number of tickets.")
        return
    total=ticketcount * price
    print("\nBooking details")
    print(f"Movie:   {moviename}")
    print(f"Time:    {showtime}")
    print(f"Tickets: {ticketcount}")
    print(f"Total:   Rs. {total}")
    confirm=input("\nConfirm booking? (yes/no): ").strip().lower()
    if confirm=="yes":
        bookings.append({
            "movie": moviename,
            "time": showtime,
            "tickets": ticketcount,
            "total": total,
        })
        print("Booking successful!")
    else:
        print("Booking cancelled.")
def show_bookings():
    if not bookings:
        print("\nNo bookings found.")
        return
    print("\nMy Bookings")
    for number, booking in enumerate(bookings, start=1):
        print(f"\nBooking {number}")
        print(f"Movie:   {booking['movie']}")
        print(f"Time:    {booking['time']}")
        print(f"Tickets: {booking['tickets']}")
        print(f"Total:   Rs. {booking['total']}")
def main():
    print("Movie Ticket Booking System")
    name=input("Enter your name: ")
    while True:
        print("\n1. Movies")
        print("2. Show Timings")
        print("3. Book Ticket")
        print("4. My Bookings")
        print("5. Exit")
        choice=input("Enter your choice: ")
        if choice=="1":
            showmovies()
        elif choice=="2":
            showtimings()
        elif choice=="3":
            bookticket()
        elif choice=="4":
            show_bookings()
        elif choice=="5":
            print(f"Thank you, {name}! Visit again!")
            break
        else:
            print("Invalid choice.")
if __name__ == "__main__":
    main()
```
